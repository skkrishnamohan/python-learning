### Series — A Single Column of Data

```python
# A Series is like a list or 1D array, but with an INDEX (labels)

import pandas as pd

# Create from a list
cpu_usage = pd.Series([45, 87, 92, 55, 78])
print(cpu_usage)
# 0    45
# 1    87
# 2    92
# 3    55
# 4    78
# dtype: int64

# Notice: each value has an INDEX (0, 1, 2, 3, 4) automatically
```

```python
# Create with CUSTOM index (labels instead of numbers)

cpu_usage = pd.Series(
    [45, 87, 92, 55, 78],
    index=['web-01', 'web-02', 'db-01', 'cache-01', 'web-03']
)
print(cpu_usage)
# web-01      45
# web-02      87
# db-01       92
# cache-01    55
# web-03      78

# Access by label
print(cpu_usage['db-01'])        # 92
print(cpu_usage['web-01'])       # 45

# Create from a dictionary (keys become the index)
memory = pd.Series({'web-01': 62, 'web-02': 78, 'db-01': 85})
print(memory)
```

```python
# Series operations — just like NumPy, operations are element-wise

cpu = pd.Series({'web-01': 45, 'web-02': 87, 'db-01': 92})

print(cpu > 80)              # Boolean Series
print(cpu[cpu > 80])         # Filter: only values above 80
print(cpu.mean())            # 74.67
print(cpu.max())             # 92
print(cpu.idxmax())          # 'db-01' -> label of max value
```

### DataFrame — A Table of Data

```python
# A DataFrame is a 2D table — like an Excel spreadsheet
# Each COLUMN is a Series
# Each ROW is one record/observation

import pandas as pd

# Most common way: create from a DICTIONARY
# Keys become column names, values become the data

data = {
    'server':   ['web-01', 'web-02', 'db-01', 'cache-01', 'web-03'],
    'cpu':      [45, 87, 92, 55, 78],
    'memory':   [62, 78, 85, 40, 71],
    'status':   ['running', 'running', 'running', 'stopped', 'running'],
    'region':   ['eastus', 'eastus', 'westeurope', 'eastus', 'westeurope']
}

df = pd.DataFrame(data)
print(df)
#     server  cpu  memory   status      region
# 0   web-01   45      62  running      eastus
# 1   web-02   87      78  running      eastus
# 2    db-01   92      85  running  westeurope
# 3  cache-01  55      40  stopped      eastus
# 4   web-03   78      71  running  westeurope

# That's a proper TABLE with rows and named columns!
```

```python
# Other ways to create DataFrames

import pandas as pd

# From a list of dictionaries (each dict = one row)
rows = [
    {'name': 'Krishna', 'age': 25, 'role': 'DevOps'},
    {'name': 'Priya',   'age': 28, 'role': 'ML Engineer'},
    {'name': 'Amit',    'age': 30, 'role': 'SRE'},
]
df2 = pd.DataFrame(rows)
print(df2)

# From a NumPy array
import numpy as np
arr = np.random.randint(50, 100, size=(4, 3))
df3 = pd.DataFrame(arr, columns=['Math', 'Science', 'English'],
                        index=['Student1', 'Student2', 'Student3', 'Student4'])
print(df3)
```

### Viewing Data — head(), tail(), info(), describe()

```python
# When working with real data, DataFrames can have THOUSANDS of rows
# You don't want to print all of them!
# These methods help you INSPECT the data quickly

import pandas as pd

data = {
    'server': ['web-01', 'web-02', 'db-01', 'cache-01', 'web-03',
               'web-04', 'db-02', 'cache-02', 'web-05', 'lb-01'],
    'cpu':    [45, 87, 92, 55, 78, 63, 88, 42, 71, 50],
    'memory': [62, 78, 85, 40, 71, 55, 90, 38, 65, 48],
    'status': ['running', 'running', 'running', 'stopped', 'running',
               'running', 'running', 'stopped', 'running', 'running']
}
df = pd.DataFrame(data)

print('--- head() --- (first 5 rows by default)')
print(df.head())          # first 5 rows
print(df.head(3))         # first 3 rows

print('\n--- tail() --- (last 5 rows by default)')
print(df.tail(3))         # last 3 rows
```

```python
# shape, info(), describe()

print('Shape:', df.shape)     # (10, 4) -> 10 rows, 4 columns
print('Columns:', df.columns.tolist())   # ['server', 'cpu', 'memory', 'status']
print('Data types:')
print(df.dtypes)              # server: object, cpu: int64, memory: int64, status: object

print('\n--- info() --- (summary of the DataFrame)')
df.info()    # shows column names, non-null counts, dtypes, memory usage

print('\n--- describe() --- (statistical summary of numeric columns)')
print(df.describe())
# Shows: count, mean, std, min, 25%, 50%, 75%, max for each numeric column

# info()     -> overview of the structure (columns, types, nulls)
# describe() -> quick statistics (mean, min, max, etc.)
```

### Selecting Columns

```python
# Selecting a SINGLE column -> returns a Series

import pandas as pd

data = {
    'server': ['web-01', 'web-02', 'db-01', 'cache-01', 'web-03'],
    'cpu':    [45, 87, 92, 55, 78],
    'memory': [62, 78, 85, 40, 71],
    'status': ['running', 'running', 'running', 'stopped', 'running']
}
df = pd.DataFrame(data)

# Method 1: bracket notation (always works)
print(df['cpu'])          # returns a Series of CPU values

# Method 2: dot notation (shorter, but doesn't work for column names with spaces)
print(df.cpu)             # same result

# Selecting MULTIPLE columns -> returns a DataFrame
print(df[['server', 'cpu']])        # double brackets! -> DataFrame
print(df[['server', 'cpu', 'status']])   # any combination of columns
```

### Selecting Rows — loc and iloc

```python
# Two main ways to select rows:
# loc  -> by LABEL (index name)
# iloc -> by INTEGER position (0, 1, 2, ...)

# loc = Location by label
# iloc = Integer location

import pandas as pd

data = {
    'server': ['web-01', 'web-02', 'db-01', 'cache-01', 'web-03'],
    'cpu':    [45, 87, 92, 55, 78],
    'memory': [62, 78, 85, 40, 71],
    'status': ['running', 'running', 'running', 'stopped', 'running']
}
df = pd.DataFrame(data)

# iloc — by position (like list indexing)
print(df.iloc[0])          # first row (as Series)
print(df.iloc[2])          # third row
print(df.iloc[0:3])        # first 3 rows (as DataFrame)
print(df.iloc[-1])         # last row

# iloc with row AND column positions
print(df.iloc[0, 1])       # row 0, column 1 -> 45 (cpu of first row)
print(df.iloc[0:3, 0:2])   # rows 0-2, columns 0-1
```

```python
# loc — by label (index name and column name)

import pandas as pd

data = {
    'cpu':    [45, 87, 92, 55, 78],
    'memory': [62, 78, 85, 40, 71],
    'status': ['running', 'running', 'running', 'stopped', 'running']
}
df = pd.DataFrame(data, index=['web-01', 'web-02', 'db-01', 'cache-01', 'web-03'])

# loc uses LABELS (the index names and column names)
print(df.loc['web-01'])              # entire row for web-01
print(df.loc['db-01', 'cpu'])        # cpu of db-01 -> 92
print(df.loc['web-01':'db-01'])      # rows from web-01 TO db-01 (inclusive!)
print(df.loc[['web-01', 'db-01'], ['cpu', 'memory']])   # specific rows + columns

# IMPORTANT: loc slicing is INCLUSIVE on both ends!
# df.loc['web-01':'db-01'] includes db-01
# df.iloc[0:3] does NOT include index 3 (standard Python slicing)
```

### Filtering Data with Conditions

```python
# Filtering = selecting rows that meet a condition
# This is one of the MOST COMMON operations in pandas

import pandas as pd

data = {
    'server': ['web-01', 'web-02', 'db-01', 'cache-01', 'web-03'],
    'cpu':    [45, 87, 92, 55, 78],
    'memory': [62, 78, 85, 40, 71],
    'status': ['running', 'running', 'running', 'stopped', 'running']
}
df = pd.DataFrame(data)

# Filter: CPU > 80
high_cpu = df[df['cpu'] > 80]
print('High CPU servers:')
print(high_cpu)

# Filter: status is 'running'
running = df[df['status'] == 'running']
print('\nRunning servers:')
print(running)

# Multiple conditions — use & (and), | (or)
# IMPORTANT: wrap each condition in parentheses!
critical = df[(df['cpu'] > 80) & (df['status'] == 'running')]
print('\nCritical (high CPU + running):')
print(critical)
```

```python
# More filtering techniques

# isin() — check if value is in a list
web_servers = df[df['server'].isin(['web-01', 'web-02', 'web-03'])]
print('Web servers only:')
print(web_servers)

# str.contains() — check if a string column contains a substring
web = df[df['server'].str.contains('web')]
print('\nServers containing "web":')
print(web)

# between() — range filter
moderate = df[df['cpu'].between(50, 80)]
print('\nCPU between 50-80:')
print(moderate)
```

### Adding, Renaming, and Dropping Columns

```python
# Adding a new column — just assign to a new column name

import pandas as pd

data = {
    'server': ['web-01', 'web-02', 'db-01', 'cache-01'],
    'cpu':    [45, 87, 92, 55],
    'memory': [62, 78, 85, 40]
}
df = pd.DataFrame(data)

# Add a column with a calculated value
df['avg_usage'] = (df['cpu'] + df['memory']) / 2
print(df)

# Add a column based on a condition
df['status'] = df['cpu'].apply(lambda x: 'CRITICAL' if x > 80 else 'OK')
print(df)

# Add a column with a constant value
df['region'] = 'eastus'
print(df)
```

```python
# Renaming columns
df_renamed = df.rename(columns={'cpu': 'cpu_percent', 'memory': 'mem_percent'})
print(df_renamed.columns.tolist())

# Dropping columns
df_dropped = df.drop(columns=['region', 'avg_usage'])
print(df_dropped)

# Dropping rows by index
df_no_first = df.drop(index=[0])      # drop first row
print(df_no_first)

# NOTE: drop() returns a NEW DataFrame — original is NOT modified
# To modify in place: df.drop(columns=['col'], inplace=True)
```

### Reading CSV Files

```python
# CSV (Comma Separated Values) is the MOST common data file format
# pandas makes reading CSVs incredibly easy

import pandas as pd

# First, let's CREATE a CSV file to practice with
data = {
    'server': ['web-01', 'web-02', 'db-01', 'cache-01', 'web-03',
               'web-04', 'db-02', 'cache-02'],
    'cpu':    [45, 87, 92, 55, 78, 63, 88, 42],
    'memory': [62, 78, 85, 40, 71, 55, 90, 38],
    'disk':   [30, 45, 78, 20, 55, 40, 82, 15],
    'status': ['running', 'running', 'running', 'stopped', 'running',
               'running', 'running', 'stopped'],
    'region': ['eastus', 'eastus', 'westeurope', 'eastus', 'westeurope',
               'eastus', 'westeurope', 'eastus']
}
df = pd.DataFrame(data)
df.to_csv('servers.csv', index=False)    # save WITHOUT the index column
print('CSV file created!')

# Now READ it back
df_loaded = pd.read_csv('servers.csv')
print(df_loaded)

# That's it! ONE line to read an entire CSV file into a DataFrame
# pd.read_csv() can also read from URLs:
# df = pd.read_csv('https://example.com/data.csv')
```

### Sorting Data

```python
# sort_values() — sort by one or more columns

import pandas as pd
df = pd.read_csv('servers.csv')

# Sort by CPU (ascending — lowest first)
print(df.sort_values('cpu'))

# Sort by CPU (descending — highest first)
print(df.sort_values('cpu', ascending=False))

# Sort by multiple columns: first by status, then by cpu
print(df.sort_values(['status', 'cpu'], ascending=[True, False]))

# NOTE: sort_values() returns a NEW DataFrame
# Original df is NOT modified unless you use inplace=True
```
