### What is a Module?

```python
# A MODULE is simply a Python file (.py) containing functions, classes, and variables
# Any .py file is a module!
#
# Three types of modules:
# 1. BUILT-IN modules — come with Python (os, sys, math, datetime, json)
# 2. THIRD-PARTY modules — installed with pip (numpy, pandas, flask, mlflow)
# 3. YOUR OWN modules — .py files you create yourself
#
# A PACKAGE is a collection of modules inside a folder
# Think of it as: Module = one file, Package = a folder of files
```

### Three Ways to Import

```python
# Way 1: import the entire module
import math

print(math.sqrt(25))        # 5.0  -> square root
print(math.pi)              # 3.141592653589793
print(math.ceil(4.2))       # 5    -> round UP
print(math.floor(4.8))      # 4    -> round DOWN

# You must use the module name prefix: math.sqrt(), math.pi
```

```python
# Way 2: from module import specific things
from math import sqrt, pi

print(sqrt(25))      # 5.0   -> no need for 'math.' prefix!
print(pi)            # 3.141592653589793

# Advantage: shorter code (no module prefix)
# Disadvantage: can cause name conflicts if two modules have same function name
```

```python
# Way 3: import with an alias (nickname)
import numpy as np          # we already know this one!
import datetime as dt

# Common aliases you'll see everywhere:
# import numpy as np
# import pandas as pd
# import matplotlib.pyplot as plt
# import datetime as dt

# These are standard conventions — everyone uses them
```

```python
# AVOID: from module import * (importing EVERYTHING)

# from math import *    <- BAD practice!
# This imports ALL functions from math into your namespace
# Problems:
# 1. You don't know which names came from where
# 2. Can accidentally override your own variables/functions
# 3. Makes code harder to read and debug
#
# ALWAYS be explicit about what you import
```

### Built-in Module: math

```python
# math module — mathematical functions

import math

print(math.sqrt(144))       # 12.0         -> square root
print(math.pow(2, 10))      # 1024.0       -> 2^10
print(math.ceil(4.1))       # 5            -> round UP
print(math.floor(4.9))      # 4            -> round DOWN
print(math.pi)              # 3.14159...   -> value of π
print(math.e)               # 2.71828...   -> Euler's number
print(math.log(100, 10))    # 2.0          -> log base 10 of 100
print(math.factorial(5))    # 120          -> 5! = 5×4×3×2×1
print(math.gcd(12, 8))      # 4            -> greatest common divisor
```

### Built-in Module: random

```python
# random module — generating random numbers and choices

import random

print(random.random())              # random float between 0.0 and 1.0
print(random.randint(1, 100))       # random integer between 1 and 100 (inclusive)
print(random.uniform(1.0, 10.0))    # random float between 1.0 and 10.0

# Pick a random item from a list
servers = ['web-01', 'web-02', 'web-03', 'web-04']
print(random.choice(servers))       # randomly picks one server

# Pick multiple random items
print(random.sample(servers, 2))    # randomly picks 2 servers (no repeats)

# Shuffle a list in place
random.shuffle(servers)
print(servers)                      # list is now in random order

# DevOps use: random jitter in retry delays, random test data,
# randomly selecting servers for health checks
```

### Built-in Module: os

```python
# os module — interact with the operating system
# This is the MOST IMPORTANT module for DevOps scripting

import os

# Current working directory
print('Current directory:', os.getcwd())

# List files and directories
print('Files here:', os.listdir('.'))     # '.' means current directory

# Check if a file/directory exists
print('sample.txt exists?', os.path.exists('sample.txt'))

# Check if it's a file or directory
print('Is file?', os.path.isfile('sample.txt'))
print('Is dir?',  os.path.isdir('sample.txt'))
```

```python
# os.path — working with file paths (cross-platform!)

import os

# Join paths safely (works on Windows AND Linux)
path = os.path.join('home', 'user', 'documents', 'report.pdf')
print(path)    # home/user/documents/report.pdf (or home\user\...on Windows)

# Get parts of a path
filepath = '/var/log/nginx/access.log'
print('Directory:', os.path.dirname(filepath))     # /var/log/nginx
print('Filename:',  os.path.basename(filepath))    # access.log
print('Extension:', os.path.splitext(filepath)[1]) # .log

# Get file size
if os.path.exists('sample.txt'):
    size = os.path.getsize('sample.txt')
    print(f'sample.txt is {size} bytes')
```

```python
# os — Environment Variables
# Environment variables are key-value pairs set in the OS
# Used for secrets, config, and system info

import os

# Read an environment variable
home = os.environ.get('USERPROFILE')    # Windows
# home = os.environ.get('HOME')         # Linux/Mac
print('Home directory:', home)

# Using .get() with a default value (safe — won't crash if missing)
db_host = os.environ.get('DB_HOST', 'localhost')    # default to 'localhost'
db_port = os.environ.get('DB_PORT', '5432')         # default to '5432'
print(f'Database: {db_host}:{db_port}')

# Set an environment variable (for current process only)
os.environ['MY_APP_ENV'] = 'development'
print('App env:', os.environ.get('MY_APP_ENV'))

# DevOps use case:
# - Read secrets from environment variables (never hardcode passwords!)
# - Check which environment we're in (dev/staging/prod)
# - Get cloud credentials (AZURE_SUBSCRIPTION_ID, AWS_ACCESS_KEY)
```

### Built-in Module: datetime

```python
# datetime module — working with dates and times
# Essential for logging, timestamps, scheduling, and reporting

from datetime import datetime, date, timedelta

# Current date and time
now = datetime.now()
print('Now:', now)                    # 2026-07-01 14:30:22.123456
print('Date only:', now.date())       # 2026-07-01
print('Time only:', now.time())       # 14:30:22.123456

# Access individual components
print(f'Year: {now.year}, Month: {now.month}, Day: {now.day}')
print(f'Hour: {now.hour}, Minute: {now.minute}, Second: {now.second}')

# Today's date
today = date.today()
print('Today:', today)                # 2026-07-01
```

```python
# Formatting dates as strings — strftime()
# 'strf' = string format time

from datetime import datetime

now = datetime.now()

# Common format codes:
# %Y = 4-digit year (2026)    %m = month (07)     %d = day (01)
# %H = hour 24h (14)          %M = minute (30)    %S = second (22)
# %A = weekday name (Monday)  %B = month name (July)

print(now.strftime('%Y-%m-%d'))                # 2026-07-01
print(now.strftime('%Y-%m-%d %H:%M:%S'))       # 2026-07-01 14:30:22
print(now.strftime('%d/%m/%Y'))                # 01/07/2026
print(now.strftime('%A, %B %d, %Y'))           # Tuesday, July 01, 2026

# Creating a log timestamp
log_timestamp = now.strftime('[%Y-%m-%d %H:%M:%S]')
print(f'{log_timestamp} Application started')
```

```python
# Date arithmetic — timedelta

from datetime import datetime, timedelta

now = datetime.now()

# Add or subtract time
tomorrow = now + timedelta(days=1)
yesterday = now - timedelta(days=1)
next_week = now + timedelta(weeks=1)
two_hours_later = now + timedelta(hours=2)

print(f'Now:        {now.strftime("%Y-%m-%d %H:%M")}')
print(f'Tomorrow:   {tomorrow.strftime("%Y-%m-%d %H:%M")}')
print(f'Yesterday:  {yesterday.strftime("%Y-%m-%d %H:%M")}')
print(f'Next week:  {next_week.strftime("%Y-%m-%d %H:%M")}')

# Calculate the difference between two dates
deployment_date = datetime(2026, 6, 15)
days_since = (now - deployment_date).days
print(f'Days since last deployment: {days_since}')
```

### if __name__ == '__main__'

```python
# This is the MOST confusing line for Python beginners
# But it's actually simple once you understand it
#
# Every Python file has a special variable called __name__
#
# When you RUN a file directly (python myfile.py):
#   __name__ is set to '__main__'
#
# When you IMPORT a file from another file (import myfile):
#   __name__ is set to the module name ('myfile')
#
# So:
#   if __name__ == '__main__':  means:
#   'Only run this code if I'm being run DIRECTLY, not imported'

print(f'In this notebook, __name__ = {__name__}')
# Output: __name__ = __main__  (because we're running it directly)
```

```python
# Why is this useful?
#
# Imagine you have a file called 'utils.py' with helper functions.
# You want to be able to:
#   1. IMPORT it from other files (to use the functions)
#   2. RUN it directly to TEST those functions
#
# Without the guard, any test code runs EVERY time someone imports your file!
#
# Example file: utils.py
# ---------------------------
# def add(a, b):
#     return a + b
#
# def multiply(a, b):
#     return a * b
#
# if __name__ == '__main__':
#     # This code ONLY runs when you do: python utils.py
#     # It does NOT run when someone does: import utils
#     print('Testing add:', add(3, 5))         # Testing add: 8
#     print('Testing multiply:', multiply(3, 5)) # Testing multiply: 15
# ---------------------------
#
# From another file:
# import utils
# result = utils.add(10, 20)   # works! test code doesn't run

# You'll see this pattern in EVERY Python project
```

### Installing Packages with pip

```python
# pip is Python's PACKAGE MANAGER
# It downloads and installs third-party libraries from PyPI
# (Python Package Index — the app store for Python libraries)

# Common pip commands (run in terminal, NOT in Python):

# Install a package:
#   pip install numpy
#   pip install pandas matplotlib
#   pip install flask==2.3.0          # install a specific version

# Upgrade a package:
#   pip install --upgrade numpy

# Uninstall a package:
#   pip uninstall numpy

# List installed packages:
#   pip list

# Show info about a package:
#   pip show numpy

# Freeze current packages into requirements.txt:
#   pip freeze > requirements.txt

# Install all packages from requirements.txt:
#   pip install -r requirements.txt
```

```python
# requirements.txt — the standard way to manage dependencies
#
# Every Python project should have a requirements.txt file
# It lists ALL external packages your project needs
#
# Example requirements.txt:
# numpy==1.24.3
# pandas>=2.0.0
# flask>=2.3.0,<3.0.0
# mlflow==2.8.0
# azure-storage-blob>=12.0.0
#
# Why this matters:
# When someone else (or a CI/CD pipeline) clones your project,
# they just run: pip install -r requirements.txt
# And ALL dependencies are installed automatically!
#
# In DevOps:
# - Dockerfiles often have: RUN pip install -r requirements.txt
# - CI/CD pipelines install dependencies this way
# - Azure ML environments use requirements.txt
```
