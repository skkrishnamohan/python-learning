### Opening Files — open() and File Modes

```python
# To work with a file, you first need to OPEN it
# Syntax: file = open('filename', 'mode')
#
# FILE MODES:
# 'r'  -> Read only (default) — file must exist
# 'w'  -> Write only — CREATES file if it doesn't exist
#                       OVERWRITES everything if it does exist!
# 'a'  -> Append — adds to the END of the file (doesn't overwrite)
# 'r+' -> Read AND Write — file must exist
#
# IMPORTANT DIFFERENCE:
# 'w' DELETES everything and starts fresh
# 'a' KEEPS everything and adds at the end
#
# Think of it this way:
# 'w' = erase the whiteboard and write new stuff
# 'a' = keep what's on the whiteboard and add more at the bottom
```

### Writing to Files

```python
# Let's start by CREATING a file and writing to it
# This will create a file called 'sample.txt' in the same directory

# Step 1: Open the file in write mode
file = open('sample.txt', 'w')    # 'w' = write mode

# Step 2: Write content
file.write('Hello from Python!\n')           # \n = newline
file.write('This is my first file.\n')
file.write('File handling is easy!\n')

# Step 3: CLOSE the file (very important!)
file.close()

print('File written successfully!')

# WHY close the file?
# 1. Frees up system resources (memory)
# 2. Ensures all data is actually written to disk
#    (Python may buffer/hold data in memory until close)
# 3. Other programs can't access the file while it's open
#
# FORGETTING to close is a common bug — we'll learn a better way soon!
```

### Reading from Files

```python
# Three ways to read a file:

# Method 1: read() — reads the ENTIRE file as one big string
file = open('sample.txt', 'r')    # 'r' = read mode
content = file.read()
print('--- read() ---')
print(content)
file.close()

# Method 2: readline() — reads ONE line at a time
file = open('sample.txt', 'r')
print('--- readline() ---')
print(file.readline())    # reads first line
print(file.readline())    # reads second line (cursor moves forward)
file.close()

# Method 3: readlines() — reads ALL lines into a LIST
file = open('sample.txt', 'r')
lines = file.readlines()
print('--- readlines() ---')
print(lines)    # ['Hello from Python!\n', 'This is my first file.\n', ...]
file.close()

# Which one to use?
# read()      -> small files, need entire content as a string
# readline()  -> processing one line at a time
# readlines() -> need all lines as a list (most common for log parsing)
```

```python
# Looping through a file — the most COMMON pattern

# You can iterate directly over a file object (line by line)
# This is memory-efficient — doesn't load the entire file at once

file = open('sample.txt', 'r')

for line in file:                     # Python reads one line at a time
    print(line.strip())               # .strip() removes the trailing \n

file.close()

# This is the BEST way to read large files
# read() loads everything into memory — bad for huge files
# Looping reads one line at a time — works for any file size
```

### The 'with' Statement — The RIGHT Way

```python
# The 'with' statement is a CONTEXT MANAGER
# It AUTOMATICALLY closes the file when the block ends
# Even if an error occurs! (no risk of forgetting to close)
#
# This is the RECOMMENDED way to handle files in Python

# OLD way (what we did above — risky):
# file = open('sample.txt', 'r')
# content = file.read()
# file.close()                   # what if an error happens before this line?

# NEW way — with statement (safe and clean):
with open('sample.txt', 'r') as file:     # 'as file' gives it a name
    content = file.read()
    print(content)
# file is automatically closed here — no file.close() needed!

# From now on, ALWAYS use 'with' for file operations
# You'll see this pattern everywhere in Python code
```

```python
# Writing with 'with' statement

with open('servers.txt', 'w') as f:
    f.write('web-01  10.0.0.1  running  CPU:45%\n')
    f.write('web-02  10.0.0.2  running  CPU:87%\n')
    f.write('db-01   10.0.0.3  running  CPU:92%\n')
    f.write('cache-01 10.0.0.4 stopped  CPU:0%\n')

print('servers.txt written!')

# Reading it back
with open('servers.txt', 'r') as f:
    for line in f:
        print(line.strip())
```

```python
# Appending to a file (adding content without overwriting)

with open('servers.txt', 'a') as f:     # 'a' = append mode
    f.write('monitor-01 10.0.0.5 running CPU:30%\n')

print('Appended new server!')

# Reading to verify
with open('servers.txt', 'r') as f:
    print(f.read())

# Notice: the original 4 lines are STILL there
# The new line was ADDED at the end
# If we had used 'w' instead of 'a', the original 4 lines would be GONE
```

### Working with JSON Files

```python
# JSON (JavaScript Object Notation) is the MOST common data format
# in APIs, config files, and web services
# JSON looks almost identical to Python dictionaries!
#
# Python has a built-in 'json' module to work with JSON

import json

# Writing a Python dict to a JSON file
deployment = {
    'app_name': 'payment-service',
    'version': 'v3.2.1',
    'environment': 'production',
    'replicas': 3,
    'ports': [8080, 8443],
    'env_vars': {
        'DB_HOST': 'postgres-svc',
        'CACHE_HOST': 'redis-svc'
    }
}

with open('deployment.json', 'w') as f:
    json.dump(deployment, f, indent=4)    # indent=4 makes it readable

print('JSON file written!')
```

```python
# Reading a JSON file back into a Python dict

import json

with open('deployment.json', 'r') as f:
    config = json.load(f)      # json.load() reads the file and returns a dict

print(type(config))            # <class 'dict'>
print(config['app_name'])      # payment-service
print(config['replicas'])      # 3
print(config['env_vars']['DB_HOST'])   # postgres-svc

# json.dump()  -> writes dict TO a file
# json.load()  -> reads FROM a file into a dict
# json.dumps() -> converts dict to a JSON STRING (not a file)
# json.loads() -> converts a JSON STRING to a dict (not a file)

# Example of dumps/loads (string, not file)
json_string = json.dumps(config, indent=2)
print(json_string)    # pretty-printed JSON string
```

```python
# Scenario: Read a log file and count errors

# First, create a sample log file
with open('app.log', 'w') as f:
    f.write('[INFO]  2026-07-01 10:00:01 | Application started\n')
    f.write('[INFO]  2026-07-01 10:00:05 | Health check passed\n')
    f.write('[ERROR] 2026-07-01 10:01:22 | Connection timeout on db-01\n')
    f.write('[WARN]  2026-07-01 10:02:15 | Memory usage at 85%\n')
    f.write('[ERROR] 2026-07-01 10:03:44 | Pod OOMKilled - out of memory\n')
    f.write('[INFO]  2026-07-01 10:04:00 | Restarting pod...\n')
    f.write('[INFO]  2026-07-01 10:04:10 | Pod restarted successfully\n')
    f.write('[ERROR] 2026-07-01 10:05:30 | API rate limit exceeded\n')

# Now read and analyse the log file
error_count = 0
errors = []

with open('app.log', 'r') as f:
    for line in f:
        line = line.strip()          # remove trailing \n
        if '[ERROR]' in line:
            error_count += 1
            errors.append(line)

print(f'Total errors found: {error_count}')
print('\nError details:')
for err in errors:
    print(f'  {err}')
```

### Data Engineering Formats: CSV & JSON Lines (JSONL)

```python
# In Data Engineering & DevOps, two formats dominate:
#   1. CSV   -> tabular data (use Python built-in `csv` module)
#   2. JSONL -> JSON Lines (1 JSON object per line) — standard for streaming logs!
#
# Why JSONL instead of a huge JSON array?
#   - You can stream-read a 10GB JSONL file line-by-line without loading it all into RAM!

import csv, json

# Reading CSV with csv.DictReader
with open("datasets/servers.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    high_cpu = [r for r in reader if float(r["cpu"]) > 90]
print(f"Servers > 90% CPU: {len(high_cpu)}")

# Streaming JSONL logs line-by-line (memory-safe!)
errors = []
with open("datasets/app_logs.jsonl", "r", encoding="utf-8") as f:
    for line in f:
        record = json.loads(line)
        if record["level"] == "ERROR":
            errors.append(record)
print(f"ERROR logs in app_logs.jsonl: {len(errors)}")
print("Sample error:", errors[0] if errors else None)
```
