### String Indexing & Slicing

```python
# Strings are SEQUENCES of characters
# Just like lists, each character has an INDEX (position number)
#
# Positive index: starts from 0 (left to right)
# Negative index: starts from -1 (right to left)

text = 'Python'
#        P  y  t  h  o  n
# index: 0  1  2  3  4  5
# neg:  -6 -5 -4 -3 -2 -1

print(text[0])      # P   -> first character
print(text[3])      # h   -> 4th character (index starts at 0!)
print(text[-1])     # n   -> last character
print(text[-3])     # h   -> 3rd from the end

print(len(text))    # 6   -> total number of characters
```

```python
# Slicing — extracting a PORTION of the string
# Syntax: string[start : stop : step]   (stop follows n-1 rule)

text = 'DevOps Engineer'

print(text[0:6])      # DevOps     -> index 0 to 5 (6-1)
print(text[7:])       # Engineer   -> index 7 to end
print(text[:6])       # DevOps     -> start to index 5
print(text[:])        # DevOps Engineer  -> full string
print(text[::2])      # DvpEgne   -> every 2nd character
print(text[::-1])     # reenignE spOveD  -> reverse the string!

# Reversing a string with [::-1] is a very common Python trick
# You'll see this in interview questions about palindromes
```

### String Immutability

```python
# Strings are IMMUTABLE — you CANNOT change them in place
# This is different from lists (lists are mutable)

name = 'Krishna'

# Trying to change a character directly -> TypeError
# name[0] = 'k'    # TypeError: 'str' object does not support item assignment

# So how do you "change" a string?
# You create a NEW string!

name = 'k' + name[1:]     # take 'k' + 'rishna' = 'krishna'
print(name)               # krishna

# Or use string methods (they all return NEW strings, never modify the original)
greeting = 'Hello World'
upper_greeting = greeting.upper()    # creates a NEW string
print(greeting)                      # Hello World  (original unchanged!)
print(upper_greeting)                # HELLO WORLD  (new string)
```

### Essential String Methods

```python
# Case methods — changing the case of strings

text = 'Hello World'

print(text.upper())       # HELLO WORLD     -> all uppercase
print(text.lower())       # hello world     -> all lowercase
print(text.title())       # Hello World     -> first letter of each word capitalised
print(text.capitalize())  # Hello world     -> only first letter of the string capitalised
print(text.swapcase())    # hELLO wORLD     -> swap upper <-> lower

# When to use:
# .lower() -> for case-insensitive comparisons:
#   if user_input.lower() == 'yes':  <- handles 'Yes', 'YES', 'yEs' etc.
# .upper() -> for log levels, status codes:
#   status = 'running'.upper()  -> 'RUNNING'
```

```python
# strip() — removing whitespace (or specified characters) from the edges

# Real-world scenario: user input often has extra spaces
# Or config file values have trailing whitespace

dirty = '   hello world   '

print(dirty.strip())       # 'hello world'     -> removes spaces from BOTH sides
print(dirty.lstrip())      # 'hello world   '  -> removes from LEFT only
print(dirty.rstrip())      # '   hello world'  -> removes from RIGHT only

# Can also strip specific characters
url = 'https://example.com/'
print(url.rstrip('/'))     # 'https://example.com'  -> removes trailing slash

# This is VERY common when reading files:
# line = file.readline().strip()   <- removes the '\n' at the end of each line
```

```python
# split() — breaking a string into a LIST of parts
# join() — combining a LIST of strings into ONE string
# These two are the MOST USED string methods in DevOps scripting

# split() — splits by whitespace by default
sentence = 'Python is awesome'
words = sentence.split()          # split by spaces (default)
print(words)                      # ['Python', 'is', 'awesome']

# split() with a specific delimiter
csv_line = 'name,age,city,country'
fields = csv_line.split(',')      # split by comma
print(fields)                     # ['name', 'age', 'city', 'country']

# Splitting a log line
log = '[ERROR] 2026-07-01 | Connection timeout on db-01'
parts = log.split(' | ')          # split by ' | '
print(parts)                      # ['[ERROR] 2026-07-01', 'Connection timeout on db-01']
```

```python
# join() — the REVERSE of split()
# Takes a list and joins them into a single string with a separator
# Syntax: 'separator'.join(list)

words = ['Python', 'is', 'awesome']
sentence = ' '.join(words)         # join with space
print(sentence)                    # Python is awesome

# Joining with different separators
servers = ['web-01', 'web-02', 'db-01', 'cache-01']
print(', '.join(servers))          # web-01, web-02, db-01, cache-01
print(' | '.join(servers))         # web-01 | web-02 | db-01 | cache-01
print('/'.join(['usr', 'local', 'bin']))   # usr/local/bin  -> building a path!

# IMPORTANT: join() only works with lists of STRINGS
# numbers = [1, 2, 3]
# ','.join(numbers)   # TypeError: sequence item 0: expected str, got int
# Fix: convert to strings first
numbers = [1, 2, 3]
print(','.join(str(n) for n in numbers))   # 1,2,3
```

```python
# replace() — replaces ALL occurrences of a substring with another

text = 'Hello World'
print(text.replace('World', 'Python'))    # Hello Python
print(text.replace('l', 'L'))             # HeLLo WorLd  -> replaces ALL 'l'

# Can limit the number of replacements with a 3rd argument
print(text.replace('l', 'L', 1))          # HeLlo World  -> replaces only FIRST 'l'

# DevOps use case: fixing config values
config_line = 'server=old-host.example.com'
updated = config_line.replace('old-host', 'new-host')
print(updated)    # server=new-host.example.com
```

```python
# find() and count() — searching within strings

text = 'Python is fun and Python is powerful'

# find() — returns the INDEX of the first occurrence
print(text.find('Python'))     # 0    -> found at index 0
print(text.find('is'))         # 7    -> found at index 7
print(text.find('Java'))       # -1   -> NOT found (returns -1, no error)

# count() — counts how many times a substring appears
print(text.count('Python'))    # 2    -> 'Python' appears twice
print(text.count('is'))        # 2    -> 'is' appears twice
print(text.count('Java'))      # 0    -> 'Java' not found

# index() is like find() BUT raises ValueError if not found
# Use find() when you're not sure if the substring exists
# Use 'in' operator for simple existence checks:
print('Python' in text)        # True
print('Java' in text)          # False
```

```python
# startswith() and endswith() — checking the beginning and end of strings

filename = 'deploy_script.py'

print(filename.endswith('.py'))      # True   -> it's a Python file
print(filename.endswith('.sh'))      # False  -> not a shell script
print(filename.startswith('deploy')) # True   -> starts with 'deploy'

# Very useful in DevOps:
# - Checking file types: if filename.endswith('.yaml'):
# - Checking log levels: if line.startswith('[ERROR]'):
# - Checking URLs: if url.startswith('https://'):

# Can also check multiple options with a tuple
print(filename.endswith(('.py', '.sh', '.yml')))   # True -> matches .py
```

```python
# Checking string content — isdigit(), isalpha(), isalnum()

print('12345'.isdigit())      # True   -> all characters are digits
print('123abc'.isdigit())     # False  -> has non-digit characters

print('hello'.isalpha())      # True   -> all characters are letters
print('hello123'.isalpha())   # False  -> has digits

print('hello123'.isalnum())   # True   -> all alphanumeric (letters + digits)
print('hello 123'.isalnum())  # False  -> space is not alphanumeric

print('   '.isspace())        # True   -> all whitespace

# Use case: validating user input
port = input('Enter port number: ')
if port.isdigit():
    print('Valid port:', int(port))
else:
    print('Invalid! Port must be a number.')
```

### f-strings (Formatted String Literals)

```python
# f-strings are the MODERN way to format strings in Python (Python 3.6+)
# Prefix the string with 'f' and put variables inside {curly braces}

# OLD way (what we did in Day 2 — concatenation)
name = 'Krishna'
age = 25
print('My name is ' + name + ' and I am ' + str(age) + ' years old.')    # messy!

# NEW way — f-strings (clean, readable, no str() needed for numbers)
print(f'My name is {name} and I am {age} years old.')    # clean!

# f-strings can contain EXPRESSIONS (calculations, method calls)
price = 1500
quantity = 3
print(f'Total: Rs {price * quantity}')         # Total: Rs 4500  -> expression inside {}
print(f'Name in uppercase: {name.upper()}')    # Name in uppercase: KRISHNA  -> method call
```

```python
# f-string formatting options — controlling how values are displayed

# Decimal places
pi = 3.14159265
print(f'Pi = {pi:.2f}')        # Pi = 3.14      -> 2 decimal places
print(f'Pi = {pi:.4f}')        # Pi = 3.1416    -> 4 decimal places

# Padding and alignment
name = 'CPU'
value = 87.5
print(f'{name:<10} : {value:>8.1f}%')    # CPU        :     87.5%
# <10 means left-align in 10 characters
# >8.1f means right-align in 8 characters with 1 decimal

# This is very useful for clean log output:
metrics = [('CPU', 87.5), ('Memory', 62.3), ('Disk', 91.0)]
for metric, val in metrics:
    status = 'CRITICAL' if val > 90 else 'OK'
    print(f'  {metric:<10} : {val:>6.1f}%  [{status}]')
```

```python
# f-string with dictionaries and objects

server = {'name': 'prod-web-01', 'ip': '10.0.0.1', 'status': 'running'}

print(f"Server {server['name']} ({server['ip']}) is {server['status']}")
# Server prod-web-01 (10.0.0.1) is running

# Multi-line f-strings
app_name = 'payment-service'
version = 'v3.2.1'
env = 'production'

banner = f"""
{'=' * 50}
  Deploying {app_name} {version}
  Environment: {env}
  Status: STARTED
{'=' * 50}
"""
print(banner)
```

### Escape Characters

```python
# Escape characters are special sequences starting with backslash (\)
# They represent characters that are hard to type directly

# \n  -> newline (go to next line)
# \t  -> tab (horizontal spacing)
# \\  -> literal backslash
# \'  -> single quote inside a single-quoted string
# \"  -> double quote inside a double-quoted string

print('Line 1\nLine 2\nLine 3')     # prints on 3 separate lines
print()
print('Name\tAge\tCity')             # tab-separated columns
print('Krishna\t25\tChennai')
print()
print('File path: C:\\Users\\Krishna\\Documents')   # literal backslashes
print()
print("He said \"Hello!\"")

# RAW STRINGS — disable escape sequences with 'r' prefix
# Useful for file paths and regex patterns
print(r'C:\Users\Krishna\Documents')   # prints literally, no escaping needed
```

### Quick Reference — All String Methods at a Glance

```python
# Quick Reference:
#
# CASE:        .upper() .lower() .title() .capitalize() .swapcase()
# TRIM:        .strip() .lstrip() .rstrip()
# SPLIT/JOIN:  .split(sep) .join(list)
# SEARCH:      .find(sub) .count(sub) .index(sub)
# REPLACE:     .replace(old, new)
# CHECK START: .startswith(prefix) .endswith(suffix)
# VALIDATE:    .isdigit() .isalpha() .isalnum() .isspace()
# FORMAT:      f'text {variable}'
#
# REMEMBER: All string methods return NEW strings (strings are immutable!)
# The original string is NEVER modified
```
