# Building CLI Tools with Typer

```python
# Day 39 — Building CLI Tools (Typer)
#
# DevOps engineers write a lot of scripts. 
# If you hardcode variables inside your script, nobody else can use it easily.
# You need to turn your scripts into Command Line Interfaces (CLIs).
#
# Example:
# Instead of editing `train.py` to change epochs to 100, you want to run:
#   python train.py --epochs 100
#
# The old way: `sys.argv` (messy) or `argparse` (lots of boilerplate).
# The modern way: `Typer` (by Sebastián Ramírez, creator of FastAPI).
# Typer uses type hints to automatically build CLIs and help menus!
#
# Today we'll cover:
#   1. Basic CLI with Typer
#   2. Arguments vs Options
#   3. Commands (Subcommands like `git commit`, `git push`)
#   4. Prompts and Colors
#
# Install: pip install typer[all]
```

### 1. Basic CLI with Typer

```python
# Because Typer runs from the command line, we'll write the code to a file
# and execute it using a shell command.

import os
os.makedirs('cli_demo', exist_ok=True)

basic_cli = '''import typer

def main(name: str):
    print(f"Hello {name}!")

if __name__ == "__main__":
    typer.run(main)
'''

with open('cli_demo/hello.py', 'w') as f:
    f.write(basic_cli)

# Note: Typer AUTOMATICALLY generates a --help menu based on your type hints!
print("Running without args:")
!python cli_demo/hello.py

print("\nRunning with --help:")
!python cli_demo/hello.py --help

print("\nRunning with arg:")
!python cli_demo/hello.py Alice
```

### 2. Arguments vs Options

```python
# In CLIs, there are two types of inputs:
# 1. Arguments: Required inputs (e.g., `git clone <URL>`)
# 2. Options: Optional flags with defaults (e.g., `git log --oneline`)
#
# In Typer:
# - If a function parameter has NO default value -> it's an Argument.
# - If a function parameter HAS a default value -> it's an Option (gets a -- flag).

args_options_cli = '''import typer

# 'model_name' is an Argument (required)
# 'epochs' is an Option (optional, defaults to 10)
def train(model_name: str, epochs: int = 10):
    print(f"Training {model_name} for {epochs} epochs...")

if __name__ == "__main__":
    typer.run(train)
'''

with open('cli_demo/train.py', 'w') as f:
    f.write(args_options_cli)

print("Help menu shows ARGUMENTS and OPTIONS:")
!python cli_demo/train.py --help

print("\nRunning with just the argument:")
!python cli_demo/train.py ResNet

print("\nRunning with argument and option:")
!python cli_demo/train.py ResNet --epochs 50
```

### 3. Multiple Commands (Subcommands)

```python
# Tools like Docker or Kubernetes have subcommands (e.g., `docker ps`, `docker run`).
# Typer makes this incredibly easy by using `app = typer.Typer()` and `@app.command()`.

multi_cli = '''import typer

app = typer.Typer()

@app.command()
def start(server_id: str):
    """Start a server by ID."""
    print(f"Starting server: {server_id}")

@app.command()
def stop(server_id: str, force: bool = False):
    """Stop a server. Use --force to kill it immediately."""
    if force:
        print(f"FORCE STOPPING server: {server_id}")
    else:
        print(f"Gracefully stopping server: {server_id}")

if __name__ == "__main__":
    app()
'''

with open('cli_demo/manage.py', 'w') as f:
    f.write(multi_cli)

print("Main Help Menu (Shows Commands):")
!python cli_demo/manage.py --help

print("\nHelp Menu for 'stop' Command:")
!python cli_demo/manage.py stop --help

print("\nRunning subcommands:")
!python cli_demo/manage.py start web-01
!python cli_demo/manage.py stop web-01 --force
```

### 4. Prompts and Colors

```python
# Typer provides utilities to prompt users (e.g., "Are you sure?") and print colors.
# Note: Prompts are interactive, so we can't fully run them automatically in Jupyter,
# but we can look at the code!

prompt_cli = '''import typer

def delete_database(db_name: str):
    # Ask for confirmation before doing something dangerous!
    delete = typer.confirm(f"Are you sure you want to delete {db_name}?")
    if not delete:
        typer.echo("Not deleting")
        raise typer.Abort() # Safely exit the CLI
    
    # Print in color using typer.secho (style + echo)
    typer.secho(f"Deleting {db_name}...", fg=typer.colors.RED, bold=True)

if __name__ == "__main__":
    typer.run(delete_database)
'''

with open('cli_demo/delete.py', 'w') as f:
    f.write(prompt_cli)

print("Created interactive CLI: cli_demo/delete.py")
print("Run it in your terminal to see the interactive prompt and colors!")
```
