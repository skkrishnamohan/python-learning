### What Are Exceptions?

```python
# Two types of errors in Python:
#
# 1. SYNTAX ERROR — code is written incorrectly (Python can't even understand it)
#    Example: missing colon, wrong indentation
#    These must be fixed BEFORE running
#
# 2. EXCEPTION — code is written correctly but something goes wrong DURING execution
#    Example: dividing by zero, accessing a missing key, file not found
#    These CAN be handled with try/except

# Common exceptions you'll see:
# ZeroDivisionError  -> 10 / 0
# TypeError           -> 'hello' + 5
# ValueError          -> int('abc')
# KeyError            -> dict['missing_key']
# IndexError          -> list[999] (index out of range)
# FileNotFoundError   -> open('nonexistent.txt')
# AttributeError      -> 'hello'.append('world') (string has no append)
# NameError           -> print(undefined_variable)

# Without exception handling, any of these will CRASH your program
# With try/except, you can catch them and decide what to do
```

### try / except — Basic Error Handling

```python
# Syntax:
# try:
#     code that MIGHT cause an error
# except:
#     code that runs IF an error occurs

# Example: Division by zero

# WITHOUT try/except (crashes):
# result = 10 / 0    # ZeroDivisionError: division by zero

# WITH try/except (handles the error):
try:
    result = 10 / 0
    print(result)
except:
    print('Error! Cannot divide by zero.')

print('Program continues running...')    # this line STILL executes!

# KEY POINT: Without try/except, the program would have crashed at line 10/0
# and 'Program continues running...' would NEVER print.
# With try/except, the error is caught, handled, and the program continues.
```

```python
# Example: Safe user input

try:
    age = int(input('Enter your age: '))
    print(f'Your age is {age}')
except:
    print('Invalid input! Please enter a number.')

# If the user types 'abc', int('abc') throws ValueError
# Our except block catches it and prints a friendly message
```

### Catching SPECIFIC Exceptions

```python
# Catching a generic 'except' is BAD practice
# Because it catches ALL errors — even ones you didn't expect
# This can HIDE bugs in your code!
#
# BEST PRACTICE: Always catch SPECIFIC exceptions

# Catching specific exceptions:
try:
    num = int(input('Enter a number: '))
    result = 100 / num
    print(f'Result: {result}')
except ValueError:
    print('Error: That is not a valid number!')
except ZeroDivisionError:
    print('Error: Cannot divide by zero!')

# Each except block handles a DIFFERENT type of error
# If user types 'abc'  -> ValueError is caught
# If user types '0'    -> ZeroDivisionError is caught
# If user types '5'    -> no error, result is printed normally
```

```python
# Accessing the error message with 'as'

try:
    data = {'name': 'Krishna', 'age': 25}
    print(data['salary'])        # KeyError: 'salary'
except KeyError as e:
    print(f'Key not found: {e}')  # Key not found: 'salary'

# The 'as e' captures the error object
# You can print it to see what went wrong
# This is very useful for logging errors in production
```

```python
# Catching multiple exceptions in one line

try:
    value = int('abc')
except (ValueError, TypeError) as e:
    print(f'Conversion error: {e}')

# Use a tuple of exception types when the same handling applies to multiple errors
```

### else and finally Blocks

```python
# Full syntax:
# try:
#     code that might fail
# except SomeError:
#     runs only if SomeError occurs
# else:
#     runs only if NO error occurred (success!)
# finally:
#     ALWAYS runs — error or no error (cleanup code)

try:
    num = int(input('Enter a number: '))
    result = 100 / num
except ValueError:
    print('Not a valid number!')
except ZeroDivisionError:
    print('Cannot divide by zero!')
else:
    print(f'Result: {result}')       # only runs if NO error
finally:
    print('Operation complete.')      # ALWAYS runs

# If user enters '5': Result: 20.0 -> Operation complete.
# If user enters '0': Cannot divide by zero! -> Operation complete.
# If user enters 'abc': Not a valid number! -> Operation complete.
#
# 'finally' always runs — perfect for cleanup (closing files, connections)
```

```python
# Practical example: Safe file reading

try:
    with open('nonexistent_file.txt', 'r') as f:
        content = f.read()
except FileNotFoundError:
    print('File not found! Please check the file path.')
else:
    print(f'File has {len(content)} characters')

# Since 'nonexistent_file.txt' doesn't exist:
# Output: File not found! Please check the file path.

# Without try/except, this would crash the entire script
# With try/except, we print a helpful message and continue
```

### Raising Exceptions with raise

```python
# Sometimes YOU want to deliberately cause an error
# Use 'raise' to throw an exception on purpose
#
# Why? To enforce rules in your code

def set_age(age):
    if age < 0:
        raise ValueError('Age cannot be negative!')
    if age > 150:
        raise ValueError('Age cannot be more than 150!')
    print(f'Age set to {age}')

set_age(25)       # Age set to 25

try:
    set_age(-5)   # this will raise ValueError
except ValueError as e:
    print(f'Invalid: {e}')    # Invalid: Age cannot be negative!

# This is how you VALIDATE data in your functions
# If the caller passes bad data, you raise an error immediately
```

### Creating Custom Exceptions

```python
# You can create your OWN exception types by inheriting from Exception
# This makes your error messages more specific and meaningful

class DeploymentError(Exception):
    """Raised when a deployment fails."""
    pass

class InsufficientResourcesError(Exception):
    """Raised when there aren't enough resources."""
    pass

def deploy(app_name, available_memory_gb):
    if available_memory_gb < 2:
        raise InsufficientResourcesError(
            f'Need at least 2GB memory, but only {available_memory_gb}GB available'
        )
    print(f'Deploying {app_name}... SUCCESS!')

# Test it
try:
    deploy('payment-service', 4)     # works fine
    deploy('user-service', 1)        # raises our custom error
except InsufficientResourcesError as e:
    print(f'DEPLOYMENT BLOCKED: {e}')

# Custom exceptions make your code professional and clear
# Instead of a generic ValueError, you get DeploymentError — much more descriptive!
```

### Best Practices

```python
# Exception Handling Best Practices:

# 1. ALWAYS catch SPECIFIC exceptions, not bare 'except:'
#    BAD:   except:
#    GOOD:  except ValueError:
#    WHY:   bare except catches everything, including KeyboardInterrupt
#           and SystemExit, which you usually DON'T want to catch

# 2. Don't use try/except for flow control
#    BAD:   try: value = mydict[key] except KeyError: value = default
#    GOOD:  value = mydict.get(key, default)

# 3. Keep the try block SMALL — only the line that might fail
#    BAD:   try: (20 lines of code) except: ...
#    GOOD:  try: (only the risky line) except: ...

# 4. Log the error, don't just ignore it
#    BAD:   except SomeError: pass   <- SILENT failure, very dangerous!
#    GOOD:  except SomeError as e: print(f'Error: {e}')

# 5. Use 'finally' for cleanup (closing files, connections)

# 6. Create custom exceptions for your domain
#    Makes errors easier to understand and handle
```
