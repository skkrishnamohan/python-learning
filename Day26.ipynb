# Matplotlib — Data Visualisation

```python
# Day 26 — Matplotlib
# Matplotlib is Python's MOST POPULAR plotting library
# It lets you create charts, graphs, and visualisations from your data
#
# Real-world analogy:
# If pandas is Excel's data part, matplotlib is Excel's CHARTS part
# You can create line graphs, bar charts, scatter plots, pie charts, and more
#
# Today we'll cover:
#   1. Basic plotting with plt.plot()
#   2. Line plots (trends over time)
#   3. Bar charts (comparisons)
#   4. Scatter plots (relationships between variables)
#   5. Histograms (distribution of data)
#   6. Pie charts (proportions)
#   7. Subplots (multiple charts in one figure)
#   8. Customisation (titles, labels, legends, colours, styles)
#
# Why visualisation matters for DevOps/MLOps:
# - Monitoring dashboards (CPU/memory trends)
# - ML model performance charts (accuracy over epochs)
# - Cost analysis reports
# - Deployment frequency and failure rate graphs
```

### Getting Started

```python
# Install: pip install matplotlib (already included in Anaconda)
# Import convention:

import matplotlib.pyplot as plt    # 'plt' is the standard alias
import numpy as np

# In Jupyter Notebook, add this to show plots INLINE (inside the notebook):
# %matplotlib inline    (usually automatic in modern Jupyter)

# The basic workflow for EVERY plot:
# 1. Create data
# 2. Call a plot function (plt.plot, plt.bar, plt.scatter, etc.)
# 3. Add labels and title
# 4. Call plt.show() to display

# Let's start with the simplest possible plot
x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]

plt.plot(x, y)             # create a line plot
plt.title('My First Plot') # add a title
plt.xlabel('X axis')       # label for x-axis
plt.ylabel('Y axis')       # label for y-axis
plt.show()                 # display the plot
```

### Line Plot — Trends Over Time

```python
# Line plots are perfect for showing how something CHANGES over time
# Examples: CPU usage over a day, model accuracy over epochs

import matplotlib.pyplot as plt
import numpy as np

# Simulating CPU usage over 24 hours
hours = np.arange(0, 24)
cpu_web01 = [30, 25, 20, 18, 15, 22, 45, 65, 78, 85, 88, 92,
             90, 87, 82, 75, 70, 68, 72, 80, 65, 50, 40, 35]
cpu_web02 = [40, 35, 30, 28, 25, 30, 50, 60, 70, 75, 78, 80,
             82, 79, 75, 68, 62, 58, 55, 60, 55, 48, 42, 38]

plt.figure(figsize=(12, 5))     # set the size of the figure (width, height)

plt.plot(hours, cpu_web01, label='web-01', color='#e74c3c', linewidth=2, marker='o', markersize=4)
plt.plot(hours, cpu_web02, label='web-02', color='#3498db', linewidth=2, marker='s', markersize=4)

plt.title('Server CPU Usage Over 24 Hours', fontsize=14, fontweight='bold')
plt.xlabel('Hour of Day', fontsize=12)
plt.ylabel('CPU Usage (%)', fontsize=12)
plt.legend()               # shows the label for each line
plt.grid(True, alpha=0.3)  # add a light grid
plt.axhline(y=80, color='red', linestyle='--', alpha=0.5, label='Critical threshold')

plt.tight_layout()         # prevent labels from being cut off
plt.show()

# Customisation options used:
# figsize     -> size of the plot
# color       -> line colour (hex or named)
# linewidth   -> thickness of the line
# marker      -> shape at data points ('o'=circle, 's'=square, '^'=triangle)
# markersize  -> size of markers
# alpha       -> transparency (0=invisible, 1=fully visible)
# linestyle   -> '--' dashed, '-' solid, ':' dotted
# axhline     -> horizontal reference line
```

### Bar Chart — Comparisons

```python
# Bar charts are great for COMPARING categories
# Examples: CPU per server, cost per team, errors per service

import matplotlib.pyplot as plt

servers = ['web-01', 'web-02', 'db-01', 'cache-01', 'web-03']
cpu = [45, 87, 92, 55, 78]

# Color based on value — green if OK, red if critical
colors = ['#2ecc71' if c < 80 else '#e74c3c' for c in cpu]

plt.figure(figsize=(10, 5))
plt.bar(servers, cpu, color=colors, edgecolor='white', linewidth=1.5)

plt.title('Current CPU Usage by Server', fontsize=14, fontweight='bold')
plt.xlabel('Server', fontsize=12)
plt.ylabel('CPU Usage (%)', fontsize=12)
plt.axhline(y=80, color='red', linestyle='--', alpha=0.5, label='Critical (80%)')
plt.legend()

# Add value labels on top of each bar
for i, v in enumerate(cpu):
    plt.text(i, v + 1, str(v) + '%', ha='center', fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
# Horizontal bar chart — useful when labels are long

import matplotlib.pyplot as plt

teams = ['Frontend Team', 'Backend Team', 'DevOps/Infra', 'ML/AI Team', 'QA Team']
costs = [4500, 8200, 6800, 9500, 2100]

plt.figure(figsize=(10, 5))
plt.barh(teams, costs, color=['#3498db', '#e74c3c', '#2ecc71', '#f39c12', '#9b59b6'])

plt.title('Monthly Cloud Cost by Team', fontsize=14, fontweight='bold')
plt.xlabel('Cost (USD)', fontsize=12)

# Add value labels
for i, v in enumerate(costs):
    plt.text(v + 100, i, f'${v:,}', va='center')

plt.tight_layout()
plt.show()
```

### Scatter Plot — Relationships

```python
# Scatter plots show the RELATIONSHIP between two variables
# Each dot is one data point
# Great for: CPU vs Memory, accuracy vs training time, cost vs usage

import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)
cpu = np.random.randint(20, 100, 30)
memory = cpu * 0.8 + np.random.normal(0, 10, 30)    # roughly correlated

plt.figure(figsize=(8, 6))
plt.scatter(cpu, memory, c=cpu, cmap='RdYlGn_r', s=100, alpha=0.7, edgecolors='white')

plt.title('CPU vs Memory Usage', fontsize=14, fontweight='bold')
plt.xlabel('CPU Usage (%)', fontsize=12)
plt.ylabel('Memory Usage (%)', fontsize=12)
plt.colorbar(label='CPU %')    # shows what the colours mean
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# c=cpu       -> colour each dot based on CPU value
# cmap        -> colour map ('RdYlGn_r' = red-yellow-green reversed)
# s=100       -> size of dots
# colorbar    -> legend showing colour scale
```

### Histogram — Distribution of Data

```python
# Histograms show the DISTRIBUTION of a single variable
# They answer: "How many values fall in each range?"
# Great for: response time distributions, score distributions

import matplotlib.pyplot as plt
import numpy as np

# Simulate API response times (normal distribution, mean=200ms, std=50ms)
np.random.seed(42)
response_times = np.random.normal(200, 50, 1000)

plt.figure(figsize=(10, 5))
plt.hist(response_times, bins=30, color='#3498db', edgecolor='white', alpha=0.7)

plt.title('API Response Time Distribution (n=1000)', fontsize=14, fontweight='bold')
plt.xlabel('Response Time (ms)', fontsize=12)
plt.ylabel('Frequency', fontsize=12)

# Add vertical lines for mean and percentiles
plt.axvline(np.mean(response_times), color='red', linestyle='--', label=f'Mean: {np.mean(response_times):.0f}ms')
plt.axvline(np.percentile(response_times, 95), color='orange', linestyle='--', label=f'P95: {np.percentile(response_times, 95):.0f}ms')
plt.axvline(np.percentile(response_times, 99), color='red', linestyle='-', label=f'P99: {np.percentile(response_times, 99):.0f}ms')

plt.legend()
plt.tight_layout()
plt.show()

# bins -> number of bars (ranges) to divide the data into
```

### Pie Chart — Proportions

```python
# Pie charts show PROPORTIONS — how a whole is divided into parts
# Best for: market share, cost breakdown, error type distribution

import matplotlib.pyplot as plt

labels = ['VMs', 'Storage', 'Database', 'Networking', 'Cache']
costs = [4500, 2200, 3800, 1500, 800]
colors = ['#3498db', '#2ecc71', '#e74c3c', '#f39c12', '#9b59b6']
explode = (0.05, 0, 0.05, 0, 0)    # slightly separate VM and DB slices

plt.figure(figsize=(8, 8))
plt.pie(costs, labels=labels, colors=colors, explode=explode,
        autopct='%1.1f%%', startangle=90, shadow=True,
        textprops={'fontsize': 12})

plt.title('Monthly Cloud Cost Breakdown', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

# autopct -> format of percentage labels ('%.1f%%' = 1 decimal)
# explode -> tuple of distances to 'pull out' each slice
# startangle -> rotation of the pie
```

### Subplots — Multiple Charts in One Figure

```python
# Subplots let you put MULTIPLE charts in one figure
# Like a dashboard with multiple panels

import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)
hours = np.arange(24)
cpu = np.random.randint(20, 95, 24)
memory = np.random.randint(30, 90, 24)
disk = np.random.randint(40, 85, 24)

# Create a figure with 2 rows × 2 columns = 4 subplots
fig, axes = plt.subplots(2, 2, figsize=(14, 8))

# Top-left: CPU line plot
axes[0, 0].plot(hours, cpu, color='#e74c3c', linewidth=2)
axes[0, 0].set_title('CPU Usage')
axes[0, 0].set_ylabel('%')
axes[0, 0].axhline(y=80, color='red', linestyle='--', alpha=0.3)
axes[0, 0].grid(True, alpha=0.3)

# Top-right: Memory line plot
axes[0, 1].plot(hours, memory, color='#3498db', linewidth=2)
axes[0, 1].set_title('Memory Usage')
axes[0, 1].set_ylabel('%')
axes[0, 1].grid(True, alpha=0.3)

# Bottom-left: Disk bar chart
axes[1, 0].bar(hours, disk, color='#2ecc71', alpha=0.7)
axes[1, 0].set_title('Disk Usage')
axes[1, 0].set_xlabel('Hour')
axes[1, 0].set_ylabel('%')

# Bottom-right: CPU vs Memory scatter
axes[1, 1].scatter(cpu, memory, c=cpu, cmap='RdYlGn_r', alpha=0.7)
axes[1, 1].set_title('CPU vs Memory')
axes[1, 1].set_xlabel('CPU %')
axes[1, 1].set_ylabel('Memory %')

fig.suptitle('Server Monitoring Dashboard — prod-web-01', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
```

### Plotting with Pandas (Built-in)

```python
# pandas has built-in plotting that uses matplotlib under the hood
# It's the FASTEST way to create quick visualisations from DataFrames

import pandas as pd
import matplotlib.pyplot as plt

data = {
    'server': ['web-01', 'web-02', 'db-01', 'cache-01', 'web-03'],
    'cpu':    [45, 87, 92, 55, 78],
    'memory': [62, 78, 85, 40, 71],
    'disk':   [30, 45, 78, 20, 55]
}
df = pd.DataFrame(data)

# Quick bar chart from DataFrame
df.plot(x='server', y=['cpu', 'memory', 'disk'], kind='bar', figsize=(10, 5),
        title='Server Resource Usage', rot=0)
plt.ylabel('Usage (%)')
plt.tight_layout()
plt.show()

# kind options: 'line', 'bar', 'barh', 'scatter', 'hist', 'pie', 'box'
```

### Saving Plots to Files

```python
# Save plots as image files (PNG, PDF, SVG)
# Use plt.savefig() BEFORE plt.show()

import matplotlib.pyplot as plt

servers = ['web-01', 'web-02', 'db-01', 'cache-01']
cpu = [45, 87, 92, 55]

plt.figure(figsize=(8, 5))
plt.bar(servers, cpu, color=['#2ecc71', '#e74c3c', '#e74c3c', '#2ecc71'])
plt.title('CPU Usage Report')
plt.ylabel('CPU %')

# Save BEFORE show()
plt.savefig('cpu_report.png', dpi=150, bbox_inches='tight')
# plt.savefig('cpu_report.pdf')    # save as PDF
# plt.savefig('cpu_report.svg')    # save as SVG (vector)

plt.show()
print('Plot saved as cpu_report.png!')

# dpi=150 -> dots per inch (higher = better quality)
# bbox_inches='tight' -> removes extra whitespace
```
